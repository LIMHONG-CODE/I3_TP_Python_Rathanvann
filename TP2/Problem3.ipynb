```python
#Step 1

students = ["Dara", 20, 3.45]
print("Name: ", students[0])
```

```python
#Step 2

name, age, gpa = students
print(name, "is", age,"years old GPA",gpa)
```

```python
#Step 3

students = [("Dara", 20, 3.45), ("Sophea", 21, 3.92), 
            ("Vibol", 19, 3.10), ("Malis", 22, 3.78)]
for name, age, gpa in students:
    print(name, "is", age, "years old GPA", gpa)
```

```python
#Step 4

ranking = sorted(students, key = lambda s:s[2], reverse = students)
print("The best student is: ", ranking[0][0])
```
