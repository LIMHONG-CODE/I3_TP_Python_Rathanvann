```python
#Step 1

shopping = ["rice", "egg", "milk"]
print('Start  :', shopping)
```

```python
#Step 2

shopping.append("bread")
shopping.insert(0, "water")
print("Add   : ", shopping)
```

```python
#Step 3

shopping.remove("egg")
last = shopping.pop()
print("Removed  :", last)
print("Now  :", shopping)
```

```python
#Step 4

print("First 2 items: ", shopping[:2])
print("Last 2 items: ", shopping[1:])
print("Reversed: ", shopping[::-1])
```

```python
#Step 5

print("Is milk in the list ? ", "milk" in shopping)
for i, item in enumerate(shopping, start = 1):
    print(i, item)
```
